```python
import numpy as np


def linear_regression(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    x_mean, y_mean = x.mean(), y.mean()
    ss_xy = np.sum((x - x_mean) * (y - y_mean))
    ss_x = np.sum((x - x_mean) ** 2)
    slope = ss_xy / ss_x
    intercept = y_mean - slope * x_mean
    return intercept, slope


def gradient_step(x, y, weight=0.0, bias=0.0, learning_rate=0.1):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    error = weight * x + bias - y
    gradient_w = np.mean(error * x)
    gradient_b = np.mean(error)
    return weight - learning_rate * gradient_w, bias - learning_rate * gradient_b

x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)
intercept, slope = linear_regression(x, y)
print("x_bar = %.4f, y_bar = %.4f" % (x.mean(), y.mean()))
print("SSxy = %.4f, SSx = %.4f" % (np.sum((x - x.mean()) * (y - y.mean())), np.sum((x - x.mean()) ** 2)))
print("beta0 = %.4f, beta1 = %.4f, y_hat(6) = %.4f" % (intercept, slope, intercept + slope * 6))
```

### Nhận xét câu 1

Dữ liệu có quan hệ tuyến tính hoàn hảo với phương trình `y = 1 + 2x`. Hệ số góc dương cho thấy khi `x` tăng một đơn vị thì `y` tăng trung bình hai đơn vị.

```python
x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([4, 6, 8, 11], dtype=float)
prediction = 2 * x + 1
print("y_hat =", prediction)
print("residual =", y - prediction)
print("MSE =", np.mean((y - prediction) ** 2))
```

### Nhận xét câu 2

MSE bằng `1.75`. Các residual đều nhỏ, nhưng mẫu cuối có sai số lớn hơn các mẫu còn lại nên ảnh hưởng nhiều hơn đến MSE.

```python
x = np.array([1, 2, 4], dtype=float)
y = np.array([3, 5, 9], dtype=float)
print("y_hat ban dau =", np.zeros_like(y))
print("w1, b1 =", gradient_step(x, y))
```

### Nhận xét câu 3

Gradient Descent cập nhật trọng số theo hướng làm giảm hàm mất mát. Sau bước đầu tiên, `w` và `b` đã dương, phù hợp với xu hướng tăng của dữ liệu.

```python
X = np.column_stack((np.ones(4), [1, 2, 3, 4], [2, 1, 2, 3]))
y = np.array([5, 6, 9, 12], dtype=float)
beta = np.linalg.pinv(X) @ y
print("X^T X =\n", X.T @ X)
print("X^T Y =", X.T @ y)
print("beta =", beta)
print("du doan (5, 2) =", np.array([1, 5, 2]) @ beta)
```

### Nhận xét câu 4

Hồi quy đa biến sử dụng đồng thời hai biến đầu vào. Pseudo-inverse cho nghiệm ổn định ngay cả khi việc tính ma trận nghịch đảo trực tiếp không thuận lợi.

```python
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 3, 5, 4, 6], dtype=float)
w1, b1 = gradient_step(x, y)
print("w1 = %.4f, b1 = %.4f, y_hat(6) = %.4f" % (w1, b1, w1 * 6 + b1))
```

### Nhận xét câu 5

Batch Gradient Descent sử dụng toàn bộ tập dữ liệu để tính gradient trong mỗi bước. Kết quả bước đầu cho biết hướng điều chỉnh của mô hình, chưa phải nghiệm tối ưu cuối cùng.

```python
points = {"A": (1, 1), "B": (2, 1), "C": (1, 3), "D": (4, 2)}
for name, (x1, x2) in points.items():
    value = 2 * x1 - x2 + 1
    print(name, "score =", value, "; class =", int(value >= 0))
print("duong quyet dinh: 2*x1 - x2 + 1 = 0")
```

### Nhận xét câu 6

Đây là bộ phân lớp tuyến tính. Dấu của hàm quyết định xác định lớp, còn đường `2*x1 - x2 + 1 = 0` là ranh giới giữa hai lớp.

```python
z = 0.5 * 2 - 1 * 1 + 0.2
probability = 1 / (1 + np.exp(-z))
print("z =", z, "; p =", probability)
print("threshold .5 =", int(probability >= .5), "; threshold .8 =", int(probability >= .8))
```

### Nhận xét câu 7

Xác suất sigmoid nằm trong khoảng 0 đến 1. Thay đổi ngưỡng phân lớp làm thay đổi quyết định cuối cùng dù xác suất của mẫu không đổi.

```python
actual = np.array([1, 0, 1])
probability = np.array([0.9, 0.2, 0.7])
losses = -(actual * np.log(probability) + (1 - actual) * np.log(1 - probability))
print("loss tung mau =", losses)
print("tong =", losses.sum(), "; trung binh =", losses.mean())
```

### Nhận xét câu 8

Cross-Entropy phạt mạnh các dự đoán tự tin nhưng sai. Loss trung bình càng nhỏ thì xác suất dự đoán càng phù hợp với nhãn thực tế.

```python
samples = [(np.array([1, 1.]), 1), (np.array([2, 1.]), 1), (np.array([1, 3.]), -1)]
weights, bias = np.zeros(2), 0.0
for features, target in samples:
    if target * (weights @ features + bias) <= 0:
        weights += target * features
        bias += target
print("w =", weights, "; b =", bias)
print("duong quyet dinh: %.1f*x1 + %.1f*x2 + %.1f = 0" % (*weights, bias))
```

### Nhận xét câu 9

Perceptron chỉ cập nhật khi mẫu bị phân lớp sai hoặc nằm đúng trên biên. Sau mỗi lần cập nhật, đường quyết định thay đổi để phân loại tốt hơn các mẫu đã gặp.

```python
samples = [(np.array([1, 1.]), 1), (np.array([2, 1.]), 1), (np.array([1, 2.]), -1), (np.array([2, 2.]), -1)]
weights, bias = np.ones(2), 0.0
misclassified = [features for features, target in samples if target * (weights @ features + bias) <= 0]
print("mau sai ban dau =", misclassified)
for features, target in samples:
    if target * (weights @ features + bias) <= 0:
        weights += target * features
        bias += target
print("sau mot epoch: w =", weights, "; b =", bias)
```

### Nhận xét câu 10

Nếu dữ liệu không tách tuyến tính hoàn toàn, Perceptron có thể tiếp tục thay đổi tham số qua các epoch. Kết quả sau một epoch chỉ là bước cập nhật ban đầu, chưa chắc là nghiệm tối ưu.
