```python
import numpy as np
matrix_a = np.array([[2.0, 1.0], [1.0, 3.0]])
vector_b = np.array([5.0, 6.0])
theta = np.linalg.solve(matrix_a, vector_b)
print("A =\n", matrix_a)
print("b =", vector_b)
print("theta =", theta)
print("Kiem tra A theta =", matrix_a @ theta)
```

```python
height = np.array([170, 155, 150, 175, 165], dtype=float)
weight = np.array([65, 50, 45, 70, 55], dtype=float)
design = np.column_stack((height, np.ones(height.size)))
theta = np.linalg.pinv(design) @ weight
prediction = design @ theta
print("theta = [slope, intercept] =", theta)
print("du doan tren tap huan luyen =", prediction)
print("MSE =", np.mean((weight - prediction) ** 2))
```

```python
x = np.array([1.0, 2.0, 4.0])
y = np.array([3.0, 5.0, 9.0])
weight, bias = 0.0, 0.0
learning_rate = 0.1
history = []

for step in range(10000):
    error = weight * x + bias - y
    loss = np.mean(error ** 2) / 2
    history.append(loss)
    gradient_w = np.mean(error * x)
    gradient_b = np.mean(error)
    weight -= learning_rate * gradient_w
    bias -= learning_rate * gradient_b
    if max(abs(learning_rate * gradient_w), abs(learning_rate * gradient_b)) < 1e-10:
        break

print("so buoc =", len(history))
print("w = %.6f, b = %.6f" % (weight, bias))
print("loss dau/cuoi = %.6f / %.6f" % (history[0], history[-1]))
```
