```python
import numpy as np


def ols(x, y):
    design = np.column_stack((np.ones(len(x)), x))
    return np.linalg.pinv(design) @ np.asarray(y, dtype=float)
```

```python
x = np.arange(1, 6, dtype=float)
y = np.array([1.2, 1.8, 2.6, 3.2, 3.8])
a0, a1 = ols(x, y)
print("x_bar = %.4f, y_bar = %.4f" % (x.mean(), y.mean()))
print("a0 = %.4f, a1 = %.4f" % (a0, a1))
print("du doan tuan 7, 9 =", a0 + a1 * np.array([7, 9]))
```

```python
prediction = 0.54 + 0.66 * x
residual = y - prediction
print("y_hat =", prediction)
print("residual =", residual)
print("SSE =", np.sum(residual ** 2))
```

```python
actual = np.array([80, 75, 90, 100.])
prediction = np.array([75, 85, 88, 95.])
error = actual - prediction
mae = np.mean(np.abs(error))
mse = np.mean(error ** 2)
print("|error| =", np.abs(error))
print("MAE =", mae, "MSE =", mse, "RMSE =", np.sqrt(mse))
```

```python
train = np.array([80, 90, 100, 110, 120.])
actual = np.array([80, 75.])
prediction = np.array([75, 85.])
rmse = np.sqrt(np.mean((actual - prediction) ** 2))
relative_mse = np.sum((actual - prediction) ** 2) / np.sum((actual - train.mean()) ** 2)
print("y_train_bar =", train.mean())
print("RelMSE =", relative_mse, "CV =", rmse / actual.mean())
```

```python
design = np.column_stack((np.ones(4), [1, 2, 3, 4]))
y = np.array([1, 3, 4, 8.], dtype=float)
beta = np.linalg.pinv(design) @ y
print("X^T X =\n", design.T @ design)
print("(X^T X)^-1 =\n", np.linalg.pinv(design.T @ design))
print("a =", beta, "; y_hat = %.4f + %.4f*x" % tuple(beta))
```

```python
design = np.column_stack((np.ones(4), [1, 2, 3, 4], [4, 5, 8, 2]))
y = np.array([1, 6, 8, 12.], dtype=float)
beta = np.linalg.pinv(design) @ y
print("a =", beta)
print("du doan (5, 6) =", np.array([1, 5, 6]) @ beta)
```

```python
x = np.array([1, 2, 3, 4.])
y = np.array([1, 4, 9, 15.])
design = np.column_stack((np.ones(4), x, x ** 2))
beta = np.linalg.pinv(design) @ y
sums = [np.sum(x), np.sum(y), np.sum(x ** 2), np.sum(x ** 3), np.sum(x ** 4), np.sum(x * y), np.sum(x ** 2 * y)]
print("cac tong =", sums)
print("[a0, a1, a2] =", beta)
print("y_hat(5) =", np.array([1, 5, 25]) @ beta)
```

```python
scores = np.array([40, 60, 80.])
probability = 1 / (1 + np.exp(-(-4 + 0.08 * scores)))
print("p =", probability)
print("class threshold .5 =", np.where(probability >= .5, "Dau", "Rot"))
```

```python
probability = np.array([0.2, 0.5, 0.8])
odds = probability / (1 - probability)
print("odds =", odds)
print("log-odds =", np.log(odds))
```

```python
names = ["A", "B", "C", "D"]
features = np.array([[60, 50], [70, 80], [40, 45], [90, 85.]])
z = -3 + 0.04 * features[:, 0] + 0.06 * features[:, 1]
probability = 1 / (1 + np.exp(-z))
for name, score, p in zip(names, z, probability):
    print(name, "z = %.4f, p = %.4f, class = %s, nhan = %s" % (score, p, "Dau" if p >= .5 else "Rot", "Co" if p >= .8 else "Khong"))
```

```python
actual = np.array([10, 15, 20, 25, 30.])
prediction = np.array([12, 14, 19, 27, 28.])
sst = np.sum((actual - actual.mean()) ** 2)
sse = np.sum((actual - prediction) ** 2)
print("y_bar =", actual.mean(), "SST =", sst, "SSE =", sse, "R2 =", 1 - sse / sst)
```

```python
print("penalty lambda*a1^2 =", 2 * 3 ** 2)
print("loss lambda=2 =", 40 + 2 * 3 ** 2)
print("loss lambda=5 =", 40 + 5 * 3 ** 2)
```

```python
coefficients = np.array([3, -2, .5])
print("sum abs =", np.abs(coefficients).sum())
print("penalty =", 2 * np.abs(coefficients).sum())
print("loss =", 40 + 2 * np.abs(coefficients).sum())
```

```python
coefficients = np.array([2, -1, 3.])
lasso_term = np.abs(coefficients).sum()
ridge_term = np.sum(coefficients ** 2)
print("LASSO term =", lasso_term)
print("Ridge term =", .5 * ridge_term)
print("total loss =", 50 + lasso_term + .5 * ridge_term)
```

```python
x = np.array([2, 3, 5, 7, 9.])
y = np.array([20, 25, 35, 45, 55.])
a0, a1 = ols(x, y)
prediction = a0 + a1 * x
residual = y - prediction
sst = np.sum((y - y.mean()) ** 2)
sse = np.sum(residual ** 2)
print("a0 = %.6f, a1 = %.6f, y_hat(10) = %.6f" % (a0, a1, a0 + a1 * 10))
print("MAE = %.6f, MSE = %.6f, RMSE = %.6f, R2 = %.6f" % (np.mean(np.abs(residual)), np.mean(residual ** 2), np.sqrt(np.mean(residual ** 2)), 1 - sse / sst))
```
